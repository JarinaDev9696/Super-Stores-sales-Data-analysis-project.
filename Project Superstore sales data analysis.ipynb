```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df["Sales"], bins=30)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

4. Sales Distribution (Histogram)

What I Found:

- The histogram shows the distribution of sales values.
- Most orders have low to medium sales.
- Only a few orders have very high sales.
- The sales data is not evenly distributed.


```python
category_sales = df.groupby("Category")["Sales"].sum()

category_sales.plot(kind="bar", figsize=(8,5))
plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

5. Category-wise Sales (Bar Chart)

What I Found:

- This chart compares total sales across different product categories.
- It helps identify which category contributes the most to total sales.
- The comparison makes it easier to understand category-wise sales performance.


```python
#region_sales
region_sales = df.groupby("Region")["Sales"].sum()

region_sales.plot(kind="bar", figsize=(8,5))
plt.title("Region-wise Sales")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

6. Category-wise Sales (Pie Chart)

What I Found:

- The pie chart shows the percentage contribution of each product category.
- It clearly compares the sales share of all categories.
- The category with the largest slice contributes the highest sales.
- The category with the smallest slice contributes the lowest sales.


```python
df.columns.tolist()
```

```python
#region_sales
region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(6,6))
plt.pie(region_sales, labels=region_sales.index, autopct="%1.1f%%")
plt.title("Region-wise Sales")
plt.show()
```

7. Region-wise Sales (Bar Chart)

What I Found:

- The chart compares sales across different regions.
- Some regions contribute more sales than others.
- This helps identify high-performing and low-performing regions.


```python
#category_sales
category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(6,6))
plt.pie(category_sales, labels=category_sales.index, autopct="%1.1f%%")
plt.title("Category-wise Sales")
plt.show()
```

8. Region-wise Sales (Pie Chart)

What I Found:

- The pie chart shows the percentage share of sales by each region.
- It provides a clear comparison of regional sales.
- It helps understand each region's contribution to total sales.


```python
#top_products
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

top_products.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Products by Sales")
plt.xlabel("Product Name")
plt.ylabel("Total Sales")
plt.show()
```

9. Top 10 Products by Sales

What I Found:

- This chart displays the top 10 products with the highest sales.
- It identifies the best-selling products in the Super Store Sales Dataset.
- These products contribute significantly to total sales.
- The business can focus on these products to improve revenue.


```python
#top_states
top_states = df.groupby("State")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_states.plot(kind="bar")
plt.title("Top 10 States by Sales")
plt.xlabel("State")
plt.ylabel("Total Sales")
plt.show()
```

##10. Top 10 States by Sales

What I Found:

- This chart displays the top 10 states with the highest sales.
- It helps identify the states contributing the most to overall sales.
- These states represent the strongest markets for the business.
- The analysis supports better marketing and sales planning.
